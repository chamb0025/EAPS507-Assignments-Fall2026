# Madhava-Leibniz π calculation


This notebook is designed to be a live demonstration of Python fundamentals applied to calculating π using the [Madhava-Leibniz series](https://en.wikipedia.org/wiki/Approximations_of_%CF%80#Middle_Ages)

The approximation is simple:

$$ \pi \approx 4 \cdot \left(1 - \frac{1}{3} + \frac{1}{5} - \frac{1}{7} + \frac{1}{9} \ldots -(-1)^n \frac{1}{2n - 1}  \right) $$

## Solving part 1: Iterations

```python
import math 

# initialize pi at 0
pi = 0

# set the number of iterations
N = 10000

# initialize an empty list store each term, so we can plot convergence later
pi_values = []

# initialize an empty list to store absolute error values for plot
error = []

# loop from 1 to N, adding terms to pi

for n in range(1,N):
    if n == 1:
        term = 1
    else:
        term = -((-1)**n) / (2*n - 1)
    pi = pi + term
    
    # store the term
    pi_values.append(4 * pi)
    error.append(abs(4 * pi - math.pi))
    
# multiply pi by four to get the final answer
pi = 4 * pi

# print the answer
print(f"pi = {pi:.7f}")
```

```python
# Plotting the error for number of iterations
import matplotlib.pyplot as plt

plt.plot(error)

# adding a title
plt.title("Absolute error of the Madhava-Leibniz Series")

# logging the x-axis
plt.xscale('log')

# labelling the x-axis and y-axis
plt.xlabel("Number of iterations")
plt.ylabel("Absolute error")
```

```python
# plotting the convergence of pi values as iterations go on
plt.plot(pi_values)

# adding a title
plt.title("Pi estimation using the Madhava-Leibniz Series")

# logging the x-axis
plt.xscale('log')

# labelling the x-axis and y-axis
plt.xlabel("Number of iterations")
plt.ylabel("Value")
```

It seems to be relatively flat and consistent error around 1000, so I would trust that 1000 iterations is going to be a pretty accurate number for most calculations needed.

## Part 2: Function

```python
def calc_pi(N):
    """This function takes an input for the number of terms to calculate pi based off of the Madhava-Leibniz series."""
    pi = 0
    for n in range(1,N):
        if n == 1:
            term = 1
        else:
            term = -((-1)**n) / (2*n - 1)
        pi = pi + term
    
        # store the term
        pi_values.append(4 * pi)
        error.append(abs(4 * pi - math.pi))
    # multiply pi by four to get the final answer
    pi = 4 * pi
    return pi

print(f"{calc_pi(10):.7f}")
print(f"{calc_pi(100):.7f}")
print(f"{calc_pi(10000):.7f}")
```

Functions can be extremely useful when developing scientific code because it allows us to simplify our main lines of code instead of rewriting the same function over and over again. It allows for easy to read code and having neatly organized code for others to read and be clearer to those who may have never seen a program like this before. It alsdo helps simplify calculations you may do multiple times. Since functions take inputs, you dont have to rewrite the code and change the number of iterations you want to do.
